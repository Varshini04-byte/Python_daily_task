Problem 6
Create a custom iterator that prints numbers from 1 to 10.

```python
class Iterator:
    def __init__(self):
        self.num = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.num <= 10:
            value = self.num
            self.num += 1
            return value
        else:
            raise StopIteration

obj = Iterator()

for i in obj:
    print(i)
```

Problem 8
Create a custom iterator that prints numbers in reverse order from
10 to 1.

```python
class Reverse:
    def __init__(self):
        self.num = 10
    def __iter__(self):
        return self
    def __next__(self):
        if self.num >= 1:
            value=self.num
            self.num-=1
            return value
        else:
            raise StopIteration
obj=Reverse()
for i in obj:
    print(i)
```

Problem 16
Given the following list:
numbers = [10, 15, 20, 25, 30, 35]
Create a generator that yields only the even numbers.

```python
numbers = [10, 15, 20, 25, 30, 35]

def even_numbers(num):
    for i in num:
        if i % 2 == 0:
            yield i

gen = even_numbers(numbers)

for num in gen:
    print(num)
```

Problem 24
Create a custom iterator that prints all vowels from the string below.
text = "Artificial Intelligence"

```python
class Vowel:
    def __init__(self, text):
        self.text=text
        self.index=0
    def __iter__(self):
        return self
    def __next__(self):
        vowels="AEIOUaeiou"

        while self.index < len(self.text):
            char=self.text[self.index]
            self.index+=1
            if char in vowels:
                return char
        raise StopIteration
text="Aritificial Intelligence"
vowels=Vowel(text)
for v in vowels:
    print(v)
```

Problem 25
Create a custom iterator that prints the Fibonacci sequence up to 10
terms.

```python
class Fibonacci:
    def __init__(self):
        self.a=0
        self.b=1
        self.count=0
    def __iter__(self):
        return self
    def __next__(self):
        if self.count<10:
            value=self.a
            self.a, self.b=self.b, self.a+self.b
            self.count += 1
            return value
        else:
            raise StopIteration
t=Fibonacci()
for num in t:
    print(num)
```

Problem 27
Create a generator that yields all prime numbers between 1 and 100.

```python
def prime_numbers():
    for num in range(2, 101):
        is_prime = True
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            yield num
for prime in prime_numbers():
    print(prime)
```

Problem 30
Create an infinite generator that yields multiples of 5.
Print only the first 15 values using next().

```python
def multiples_of_five():
    num = 5
    while True:
        yield num
        num += 5
gen = multiples_of_five()
for i in range(15):
    print(next(gen))
```
