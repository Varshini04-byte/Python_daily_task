1. Create a Calculator class that can add 2 numbers or 3 numbers using the same method
name.
Concept
Method Overloading (Python uses default arguments or *args to achieve overloading.)
What to do
• Create a class Calculator.
• Create one method add().
• The method should work for either 2 numbers or 3 numbers.
How to solve
1. Create class Calculator.
2. Define method add().
3. Use default arguments or *args.
4. Check how many numbers are provided.
5. Return the sum.

```python
class Calculator:
    def add(self,a,b,c=0):
        return a+b+c
a=Calculator()
print(a.add(4,5))
print(a.add(2,3,6))
    
```

2. Create an Area class that calculates the area of a square and rectangle using the same
method.
Concept
Method Overloading using default arguments.
What to do
• One parameter → Square.
• Two parameters → Rectangle.
How to solve
1. Create class Area.
2. Define method calculate().
3. If only one value is provided, calculate the square's area.
4. If two values are provided,calculate the rectangle's area.
5. Return the result.

```python
class Square:
    def area(self,a):
        return a*a
class Rectangle:
    def area(self,length,width=0):
        return length*width
a=Square()
print("The area of square:",a.area(4))
b=Rectangle()
print("The area of rectangle:",b.area(2,3))
    
```

4. Create a MathOperations class that multiplies any number of values.
Concept
Method Overloading using *args
What to do
Multiply:
• 2 numbers
• 3 numbers
• 4 numbers
• More numbers
How to solve
1. Create class MathOperations.
2. Define method multiply(*args).
3. Initialize the answer as 1.
4. Loop through all numbers.
5. Multiply each number.
6. Return the final product.

```python
class MathOperations:
    def multiply(self,*args):
        result = 1
        for num in args:
            result *= num
        return result
a=MathOperations()
print(" two numbers:",a.multiply(2,3))
print(" three numbers:",a.multiply(2,3,4))
print(" four numbers:",a.multiply(2,3,6,7))
print(" more numbers:",a.multiply(2,3,6,7,8,4,5))
```

5. Create a Display class that prints different messages based on the argument type.
Concept
Method Overloading + Type Checking
What to do
Accept:
• Integer
• String
• List
How to solve
1. Create class Display.
2. Create method show().
3. Use isinstance().
4. Print different messages depending on the object's type.

```python
class Display:
    def show(self,data):
        if isinstance(data, int):
            print("Integer:", data)
        elif isinstance(data, str):
            print("String:", data)
        elif isinstance(data, list):
            print("List:", data)
        else:
            print("Invalid data")
d=Display()
d.show(10)
d.show("Python Programming")
d.show([1,2,3,4])
d.show(23.8)
```

6. Create a Volume class that calculates the volume of a cube and cuboid.
Concept
Method Overloading
What to do
• One parameter → Cube
• Three parameters → Cuboid
How to solve
1. Create class Volume.
2. Create method calculate().
3. If one value is given, calculate the cube's volume.
4. Otherwise, calculate the cuboid's volume.

```python
class Volume:
    def calculate(self,a,b=0,c=0):
        if b==0 and c==0:
            return a*a*a
        else:
            return a*b*c
a=Volume()
print("Volume of Cube:",a.calculate(3))
print("Volume of Cuboid:",a.calculate(1,2,3))

        
```

15. Create a Point class and overload the + operator.
Concept
Operator Overloading (__add__)
What to do
Add two points together.
How to solve
1. Create class Point.
2. Store x and y coordinates.
3. Overload __add__().
4. Return a new point with added coordinates.

```python
class Point:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def __add__(self,other):
        return Point(self.x+other.x,self.y+other.y)
    def display(self):
        print("Point:",self.x,self.y)
p1=Point(1,2)
p2=Point(3,4)
p3=p1+p2
p3.display()
```

18. Create a Rectangle class and overload the == operator.

Equality Operator Overloading (__eq__)
What to do
Compare rectangles using their area.
How to solve
1. Create class Rectangle.
2. Store length and width.
3. Calculate area.
4. Overload __eq__().
5. Compare the areas.

```python
class Rectangle:
    def __init__(self,length,width):
        self.length=length
        self.width=width
    def __eq__(self,other):
        return self.length * self.width == other.length * other.width
r1=Rectangle(2,3)
r2=Rectangle(2,3)
print(r1==r2)
        
```

19. Create a Money class and overload the + and - operators.
Concept
Multiple Operator Overloading
What to do
Perform addition and subtraction of money amounts.
How to solve
1. Create class Money.
2. Store the amount.
3. Overload __add__().
4. Overload __sub__().
5. Return new Money objects after each operation.

```python
class Money:
    def __init__(self,amt):
        self.amt=amt
    def __add__(self,other):
        return Money(self.amt+other.amt)
    def __sub__(self,other):
         return Money(self.amt-other.amt)
    def display(self):
        print("amount:",self.amt)
        
m1=Money(200)
m2=Money(300)
r1=m1+m2
r2=m1-m2
r1.display()
r2.display()
```
