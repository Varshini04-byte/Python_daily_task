23. Create abstract class Shape with abstract method area(). Implement it in Circle and
Rectangle using super().

```python
from abc import ABC, abstractmethod 
class Shape(ABC):
    def __init__(self):
        print("Shape")
    @abstractmethod
    def area(self):
        pass
class Circle(Shape):
    def __init__(self,radius):
        super().__init__()
        self.radius=radius
    def area(self):
        print(3.14*self.radius*self.radius)
class Rectangle(Shape):
    def __init__(self,length,width):
        super().__init__()
        self.length=length
        self.width=width
    def area(self):
        print(self.length*self.width)
c=Circle(5)
c.area()
r=Rectangle(4,5)
r.area()
```

```python
text=input("Enter a string")
vowels=0
consonants=0
digits=0
special=0
for ch in text:
    if ch.isalpha():
        if ch.lower() in "aeiou":
            vowels += 1
        else:
            consonants += 1
    elif ch.isdigit():
        digits += 1
    elif not ch.isspace():
        special += 1
print("Vowels:",vowels)
print("Consonants:",consonants)
print("Digits:", digits)
print("Special Characters:", special)
```

• Add deposit and withdraw with validation
• Maintain a class attribute total_accounts
• Add a class method to show total accounts
• Overload + operator to transfer money from one account to another

```python
class BankAccount:
    total_accounts = 0
    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance
        BankAccount.total_accounts += 1
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(amount, "deposited successfully.")
        else:
            print("Invalid deposit amount.")
    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            print(amount, "withdrawn successfully.")
        else:
            print("Insufficient balance or invalid amount.")
    def display(self):
        print(self.name, "Balance:", self.__balance)
    @classmethod
    def show_total_accounts(cls):
        print("Total Accounts:", cls.total_accounts)
    def __add__(self, other):
        amount = int(input("Enter amount to transfer: "))
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            other.__balance += amount
            print("Transfer Successful!")
        else:
            print("Transfer Failed! Insufficient balance.")
        return other
acc1 = BankAccount("varshini", 12000)
acc2 = BankAccount("ammu", 1000)
print("Before Transactions:")
acc1.display()
acc2.display()
acc1.deposit(2000)
acc2.withdraw(2000)
acc1 + acc2
print("\nAfter Transactions:")
acc1.display()
acc2.display()
BankAccount.show_total_accounts()
```

5. Create abstract class Shape with abstract methods area() and perimeter().
Implement Circle and Rectangle using super(). Add a class attribute to count total
shapes created.

```python
from abc import ABC, abstractmethod
class Shape(ABC):
    total_shapes = 0  
    def __init__(self):
        Shape.total_shapes += 1
    @abstractmethod
    def area(self):
        pass
    @abstractmethod
    def perimeter(self):
        pass
    @classmethod
    def show_total_shapes(cls):
        print("Total Shapes Created:", cls.total_shapes)
class Circle(Shape):
    def __init__(self, radius):
        super().__init__()      
        self.radius = radius
    def area(self):
        return 3.14 * self.radius * self.radius
    def perimeter(self):
        return 2 * 3.14 * self.radius
class Rectangle(Shape):
    def __init__(self, length, width):
        super().__init__()      
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width
    def perimeter(self):
        return 2 * (self.length + self.width)
c = Circle(8)
r = Rectangle(1, 6)
print("Circle Area:", c.area())
print("Circle Perimeter:", c.perimeter())
print("Rectangle Area:", r.area())
print("Rectangle Perimeter:", r.perimeter())
Shape.show_total_shapes()
```

```python
class A:
    def show(self):
        print("This is class A")
class B(A):
    def show(self):
        print("This is class B")
        super().show()
class C(A):
    def show(self):
        print("This is class C")
        super().show()
class D(B, C):
    def show(self):
        print("This is class D")
        super().show()
class Demo:
    def show(self):
        print("This is Demo class")
def display(obj):
    obj.show()
d = D()
demo = Demo()
print("Method Resolution Order (MRO):")
print(D.mro())
print("\nCalling show() using MRO:")
d.show()
print("\nDuck Typing:")
display(d)     
display(demo)
```

9. Create a class ComplexNumber and overload +, -, *, and == operators completely.

```python
class ComplexNumber:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag
    def __add__(self, other):
        return ComplexNumber(
            self.real + other.real,
            self.imag + other.imag
        )
    def __sub__(self, other):
        return ComplexNumber(
            self.real - other.real,
            self.imag - other.imag
        )
    def __mul__(self, other):
        real = (self.real * other.real) - (self.imag * other.imag)
        imag = (self.real * other.imag) + (self.imag * other.real)
        return ComplexNumber(real, imag)
    def __eq__(self, other):
        return self.real == other.real and self.imag == other.imag
    def display(self):
        print(f"{self.real} + {self.imag}i")
c1 = ComplexNumber(4, 3)
c2 = ComplexNumber(2, 1)
add_result = c1 + c2
sub_result = c1 - c2
mul_result = c1 * c2
print("First Complex Number:")
c1.display()
print("Second Complex Number:")
c2.display()
print("\nAfter Addition:")
add_result.display()
print("After Subtraction:")
sub_result.display()
print("After Multiplication:")
mul_result.display()
print("\nAre c1 and c2 equal?", c1 == c2)
```

10. Create a class Student with private marks list.
• Add setter with validation (each mark 0-100)
• Calculate average
• Overload > operator to compare two students based on average

```python
class Student:
    def __init__(self, marks):
        self.__marks = []
        self.set_marks(marks)
    def set_marks(self, marks):
        valid = True
        for mark in marks:
            if mark < 0 or mark > 100:
                valid = False
                break
        if valid:
            self.__marks = marks
        else:
            print("Marks should be between 0 and 100.")
    def get_marks(self):
        return self.__marks
    def average(self):
        total = 0
        for mark in self.__marks:
            total += mark
        return total / len(self.__marks)
    def __gt__(self, other):
        return self.average() > other.average()
    def display(self):
        print("Marks:", self.__marks)
        print("Average:", self.average())
s1 = Student([90, 95, 89])
s2 = Student([85, 90, 80])
print("Student 1:")
s1.display()
print("\nStudent 2:")
s2.display()
if s1 > s2:
    print("\nStudent 1 has a higher average.")
else:
    print("\nStudent 2 has a higher average.")
```
