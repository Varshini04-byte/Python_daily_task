10. Create a class Student with private attribute __marks. Add getter and setter with
validation (0–100).

```python
class Student:
    def __init__(self, marks):
        self.__marks = 0
        self.set_marks(marks)
    def get_marks(self):
        return self.__marks

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Marks should be between 0 and 100.")

s = Student(85)
print("Marks:", s.get_marks())
s.set_marks(95)
print("Updated Marks:", s.get_marks())
s.set_marks(120)
print("Final Marks:", s.get_marks())
```

11. Create class BankAccount with private __balance. Add deposit and withdraw
methods. Raise custom exception if balance becomes negative.

```python
class InsufficientBalanceError(Exception):
    pass
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount
        print("Deposited:", amount)
        print("Balance:", self.__balance)

    def withdraw(self, amount):
        if self.__balance - amount < 0:
            raise InsufficientBalanceError("Balance cannot be negative.")
        self.__balance -= amount
        print("Withdrawn:", amount)
        print("Balance:", self.__balance)

account = BankAccount(5000)
account.deposit(1000)
try:
    account.withdraw(2000)
    account.withdraw(5000)
except InsufficientBalanceError as e:
    print(e)
```

```python
class Rectangle:
    def __init__(self, length, width):
        self.__length = length
        self.__width = width
    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)
r = Rectangle(10, 5)
print(r.area())
print(r.perimeter())
```

15. Create single level inheritance: Person → Student. Use super() in constructor.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
class Student(Person):
    def __init__(self, name, age, course):
        super().__init__(name, age)   # Call parent constructor
        self.course = course
    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Course:", self.course)
s = Student("Varshini", 20, "DS")
s.display()
```

18. Create abstract class Shape with abstract method area(). Implement it in Circle
and Square using super().

```python
from abc import ABC, abstractmethod
class Shape(ABC):
    def __init__(self):
        print("Shape Constructor")
    @abstractmethod
    def area(self):
        pass
class Circle(Shape):
    def __init__(self, radius):
        super().__init__()   
        self.radius = radius
    def area(self):
        return 3.14 * self.radius * self.radius
class Square(Shape):
    def __init__(self, side):
        super().__init__()   
        self.side = side
    def area(self):
        return self.side * self.side
c = Circle(5)
s = Square(4)
print( c.area())
print(s.area())
```

20. Create a class Calculator that can add 2 or 3 numbers (method overloading
using *args).

```python
class Calculator:
    def add(self, *args):
        if len(args) == 2:
            print(args[0] + args[1])
        elif len(args) == 3:
            print(args[0] + args[1] + args[2])
        else:
            print("Enter only 2 or 3 numbers.")
c = Calculator()
c.add(10, 20)
c.add(10, 20, 30)
```

21. Create class Point and overload + and > operators.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)
    def __gt__(self, other):
        return (self.x + self.y) > (other.x + other.y)
    def display(self):
        print("Point(", self.x, ",", self.y, ")")
p1 = Point(2, 3)
p2 = Point(4, 5)
p3 = p1 + p2
print("After Addition:")
p3.display()
if p1 > p2:
    print("p1 is greater than p2")
else:
    print("p2 is greater than or equal to p1")
```

25. Create a menu-driven program for file operations: Write, Read, Append, and Exit.
Handle FileNotFoundError.

```python
while True:
    print("\n1. Write")
    print("2. Read")
    print("3. Append")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        file = open("data.txt", "w")
        text = input("Enter text: ")
        file.write(text)
        file.close()
        print("written successfully.")
    elif choice == "2":
        try:
            file = open("data.txt", "r")
            print("\nFile Content:")
            print(file.read())
            file.close()
        except FileNotFoundError:
            print("File not found.")
    elif choice == "3":
        file = open("data.txt", "a")
        text = input("Enter text to append: ")
        file.write("\n" + text)
        file.close()
        print("Added successfully.")
    elif choice == "4":
        print("Exit")
        break
    else:
        print("Invalid choice")
```
