20. Class Name:
BankAccount → SavingsAccount → RecurringDeposit
Problem
Create three levels of inheritance.
Explanation
Each class should add new functionality.
Example:
• deposit
• interest
• recurring deposit maturity

```python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)
class Savings_Account(BankAccount):
    def __init__(self, balance, rate):
        super().__init__(balance)
        self.rate = rate
    def calculate_interest(self):
        interest = (self.balance * self.rate) / 100
        print("Interest:", interest)
class Recurring_Deposit(Savings_Account):
    def __init__(self, balance, rate, months, monthly_deposit):
        super().__init__(balance, rate)
        self.months = months
        self.monthly_deposit = monthly_deposit
    def calculate_maturity(self):
        total_deposit = self.months * self.monthly_deposit
        interest = (total_deposit * self.rate) / 100
        maturity_amount = total_deposit + interest
        print("Total Deposit:", total_deposit)
        print("Maturity Amount:", maturity_amount)
r = Recurring_Deposit(5000, 5, 12, 1000)
r.deposit(2000)
r.calculate_interest()
r.calculate_maturity()
```

29. Class Name:
Shape → TwoDShape → Triangle
Problem
Create multi-level inheritance.
Explanation
Triangle should calculate its area using Heron's Formula.
Store:
• side1
• side2
• side3
Implement the area calculation.

```python
class Shape:
    def area(self):
        print("Area of shape")
class TwoDShape(Shape):
    def __init__(self, side1, side2, side3):
        self.side1 = side1
        self.side2 = side2
        self.side3 = side3
class Triangle(TwoDShape):
    def area(self):
        s = (self.side1 + self.side2 + self.side3) / 2
        area = (s * (s - self.side1) * 
                (s - self.side2) * 
                (s - self.side3)) ** 0.5
        print("Area of Triangle:", area)
t = Triangle(3, 4, 5)
t.area()
```
