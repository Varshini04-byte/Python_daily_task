2. Create Person, Employee, Student, and TeachingAssistant using hybrid inheritance.
Concept
Real-world Hybrid Inheritance
What to do
• Person is the base class.
• Employee inherits Person.
• Student inherits Person.
• TeachingAssistant inherits Employee and Student.
How to solve
Hierarchy
Person
/ \
Employee Student
\ /
TeachingAssistant
Steps
1. Store name in Person.
2. Employee stores salary.
3. Student stores marks.
4. TeachingAssistant stores department.
5. Display all information.

```python
#Task2:
class person:
    def __init__(self, name, age, **kwargs):
        super().__init__(**kwargs)
        self.name = name
        self.age = age
class employee(person):
    def __init__(self, name, age, salary, **kwargs):
        super().__init__(name=name, age=age, **kwargs)
        self.salary = salary
class student(person):
    def __init__(self, name, age, marks, **kwargs):
        super().__init__(name=name, age=age, **kwargs)
        self.marks = marks
class TeachingAssistant(employee, student):
    def __init__(self, name, age, salary, marks, dept):
        super().__init__(
            name=name,
            age=age,
            salary=salary,
            marks=marks
        )
        self.dept = dept

    def display(self):
        print("Name :", self.name)
        print("Age :", self.age)
        print("Salary :", self.salary)
        print("Marks :", self.marks)
        print("Dept :", self.dept)
t = TeachingAssistant("varshini", 20, 60000, 90, "administrtive")
t.display()
```

6. Create a banking system using hybrid inheritance.
Concept
Hybrid + Real-world Example
What to do
Account
|
Savings

Loan

SavingsLoanAccount
How to solve
1. Account stores account number.
2. Savings stores balance.
3. Loan stores loan amount.
4. SavingsLoanAccount inherits both.
5. Print all details.

```python
#Task6:
class Account:
    def __init__(self,acc_no,**Kwargs):
        super().__init__(**Kwargs)
        self.__acc_no = acc_no
    def get_acc_no(self):
        return self.__acc_no
class Savings(Account):
    def __init__(self,balance,**kwargs):
        super().__init__(**kwargs)
        self.__balance = balance
    def get_balance(self):
        return self.__balance
    @staticmethod
    def ValidBalance(balance):
        return balance>=0
    def set_balance(self,balance):
        if Savings.ValidBalance(balance):
            self.__balance=balance
        else:
            print("Invalid balance")
class Loan:
    def __init__(self,amount,**kwargs):
        super().__init__(**kwargs)
        self.__amount = amount
    def get_amount(self):
        return self.__amount
    @staticmethod
    def ValidAmount(amount):
        return amount>=0
    def set_amount(self,amount):
        if Loan.ValidAmount(amount):
            self.__amount=amount
        else:
            print("Invalid amount")
class SavingsLoanAccount(Savings,Loan):
    def __init__(self,acc_no,balance,amount):
        super().__init__(
            acc_no=acc_no,
            balance=balance,
            amount=amount
        )

    def display(self):
        print("acc_no :", self.get_acc_no())
        print("balance :", self.get_balance())
        print("amount :", self.get_amount())
t = SavingsLoanAccount(123420, 60000, 30000)
t.display()
```

19. Create a hybrid inheritance structure and apply Duck Typing.
Concept
Hybrid + Duck Typing
What to do
Create hybrid inheritance.
Create another unrelated class with same method.
How to solve
1. Create hybrid hierarchy.
2. Final child has method show().
3. Another unrelated class also has show().
4. One function calls show() for both.

```python
class Parent:
    def parent_method(self):
        print("Parent Method")
class Child1(Parent):
    def child1_method(self):
        print("Child1 Method")
class Child2:
    def child2_method(self):
        print("Child2 Method")
class HybridChild(Child1, Child2):
    def show(self):
        print("HybridChild Show Method")
class AnotherClass:
    def show(self):
        print("AnotherClass Show Method")
def display(obj):
    obj.show()


h = HybridChild()
a = AnotherClass()

display(h)
display(a)
```

20. Create Employee hierarchy using Hybrid Inheritance and use Duck Typing.
Concept
Hybrid Inheritance + Duck Typing
What to do
Hierarchy
Person
|
Employee

Manager

HR

TeamLead
Another class
Freelancer
Both TeamLead and Freelancer have
work()
How to solve
1. Build hybrid inheritance.
2. Final child defines work().
3. Freelancer also defines work().
4. Write a function:
def do_work(obj):
obj.work()
5. Pass both objects to the function.
6. Observe that the same function works for unrelated classes because of Duck Typing.

```python
class Person:
    def person_info(self):
        print("Person Information")
class Employee(Person):
    def employee_info(self):
        print("Employee Information")
class Company:
    def company_info(self):
        print("Company Information")
class Manager(Employee, Company):
    def work(self):
        print("Manager is working")
class Freelancer:
    def work(self):
        print("Freelancer is working")
def do_work(obj):
    obj.work()
m = Manager()
f = Freelancer()
do_work(m)
do_work(f)
```
